```python
# --- Key Statistics for Search Visibility Decay ---

# Overall Dataset Scale
total_content_pieces = len(df)
num_clients = df['client_id'].nunique() if 'client_id' in df.columns else 'N/A'
print(f"Total content pieces tracked: {total_content_pieces:,}")
print(f"Total unique clients: {num_clients}")

print("\n--- Content Performance Trends (based on 'trend_pct') ---")

if 'trend_pct' in df.columns:
    # Categorize trends
    negative_trend_df = df[df['trend_pct'] < 0]
    positive_trend_df = df[df['trend_pct'] > 0]
    neutral_trend_df = df[df['trend_pct'] == 0]

    # Calculate counts and percentages
    num_negative_trend = len(negative_trend_df)
    pct_negative_trend = (num_negative_trend / total_content_pieces) * 100
    avg_negative_trend_pct = negative_trend_df['trend_pct'].mean() if num_negative_trend > 0 else 0.0

    num_positive_trend = len(positive_trend_df)
    pct_positive_trend = (num_positive_trend / total_content_pieces) * 100

    num_neutral_trend = len(neutral_trend_df)
    pct_neutral_trend = (num_neutral_trend / total_content_pieces) * 100

    print(f"Content with negative trend_pct (losing visibility): {num_negative_trend:,} ({pct_negative_trend:.2f}%)")
    print(f"  Average trend_pct for decaying content: {avg_negative_trend_pct:.2f}%")
    print(f"Content with positive trend_pct (gaining visibility): {num_positive_trend:,} ({pct_positive_trend:.2f}%)")
    print(f"Content with neutral trend_pct (stable visibility): {num_neutral_trend:,} ({pct_neutral_trend:.2f}%)")

    # Optionally, look at the bottom X percentile of trend_pct to identify the worst performers
    bottom_10_percentile_trend = df['trend_pct'].quantile(0.10)
    worst_performing_content = df[df['trend_pct'] <= bottom_10_percentile_trend]
    print(f"\n--- Identifying Worst Performers ---")
    print(f"Bottom 10% of content (by trend_pct) consists of {len(worst_performing_content):,} pieces.")
    print(f"  Their trend_pct is {bottom_10_percentile_trend:.2f}% or worse.")

else:
    print("Warning: 'trend_pct' column not found, cannot calculate trend statistics.")
```

### Why these statistics matter:

These numbers are critical to help us understand why a content is declining.

Total content pieces and unique clients - Shows overall volume of content that we are analysing

Percentage of content with negative 'trend_pct' - Shows how much content is actively losing search visibilty.

Average 'trend_pct' for decaying content - This metric represents the severity of the decay. A large negative percentage suggests more aggresive drops in traffic.

Bottom 10% of content - These are the pages likely in most urgent need of rewrite to prevent permanent traffic loss.
